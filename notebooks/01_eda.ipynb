# Exploratory Data Analysis (EDA)
## Heart Disease UCI Dataset

This notebook performs comprehensive EDA on the Heart Disease dataset including:
- Data loading and cleaning
- Missing value handling
- Descriptive statistics
- Feature distributions
- Correlation analysis
- Class balance analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

### 1. Load Data

```python
# Download data if not present
import sys
sys.path.insert(0, '..')
from data.download_data import download_heart_disease_dataset

df = download_heart_disease_dataset()
print(f"Dataset shape: {df.shape}")
df.head()
```

### 2. Data Quality

```python
# Missing values
print("Missing values:")
print(df.isnull().sum())
print(f"\nTotal missing: {df.isnull().sum().sum()}")

# Data types
print("\nData types:")
print(df.dtypes)
```

### 3. Descriptive Statistics

```python
df.describe()
```

### 4. Target Distribution

```python
# Class balance
target_counts = df['target'].value_counts()
print("Class distribution:")
print(target_counts)
print(f"\nClass balance ratio: {target_counts[1] / target_counts[0]:.2f}")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

target_counts.plot(kind='bar', ax=axes[0], color=['green', 'red'])
axes[0].set_title('Class Distribution')
axes[0].set_ylabel('Count')
axes[0].set_xticklabels(['No Disease (0)', 'Disease (1)'], rotation=0)

target_counts.plot(kind='pie', ax=axes[1], autopct='%1.1f%%', colors=['green', 'red'])
axes[1].set_title('Class Balance')
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('../screenshots/01_class_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

### 5. Feature Distributions

```python
# Numerical features histogram
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns[:-1]

fig, axes = plt.subplots(4, 3, figsize=(15, 12))
axes = axes.ravel()

for idx, feature in enumerate(numerical_features):
    axes[idx].hist(df[feature], bins=30, edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'{feature} Distribution')
    axes[idx].set_xlabel(feature)
    axes[idx].set_ylabel('Frequency')

plt.tight_layout()
plt.savefig('../screenshots/02_feature_distributions.png', dpi=300, bbox_inches='tight')
plt.show()
```

### 6. Correlation Analysis

```python
# Correlation matrix
correlation_matrix = df.corr()

# Heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8}, fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig('../screenshots/03_correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()

# Top correlations with target
print("\nTop 10 features correlated with target:")
target_corr = correlation_matrix['target'].sort_values(ascending=False)
print(target_corr)
```

### 7. Feature vs Target Analysis

```python
# Box plots for top correlated features
top_features = target_corr[1:7].index.tolist()

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

for idx, feature in enumerate(top_features):
    sns.boxplot(data=df, x='target', y=feature, ax=axes[idx], palette='Set2')
    axes[idx].set_title(f'{feature} vs Target')
    axes[idx].set_xlabel('Heart Disease (0=No, 1=Yes)')

plt.tight_layout()
plt.savefig('../screenshots/04_feature_target_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Summary
